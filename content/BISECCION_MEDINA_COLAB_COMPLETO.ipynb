**Regla:** si `f(XL) × f(XR) < 0`, entonces `XU = XR`; si es mayor que cero, `XL = XR`; si es cero, la raíz es exacta.

## 1. Programación del método
Este código sustituye las fórmulas `SI` de Excel por decisiones `if / else` de Python. **En rojo** aparece la columna XL o XU que tomó el XR anterior.

```python
import math
from IPython.display import display, HTML

def biseccion(f, xl, xu, tolerancia=0.005, max_iter=100):
    if f(xl) * f(xu) >= 0:
        raise ValueError("El intervalo debe tener cambio de signo (f(XL)*f(XU) < 0).")
    filas = []
    anterior = None
    cambio = None
    for i in range(1, max_iter + 1):
        xr = (xl + xu) / 2
        fxl, fxr = f(xl), f(xr)
        producto = fxl * fxr
        ea = None if anterior is None else (abs((xr - anterior) / xr) * 100 if xr != 0 else None)
        paro = producto == 0 or (ea is not None and ea < tolerancia)
        filas.append((i, xl, xu, xr, fxl, fxr, producto, ea, "OK" if paro else "", cambio))
        if paro:
            break
        if producto < 0:
            xu = xr
            cambio = "XU"
        else:
            xl = xr
            cambio = "XL"
        anterior = xr
    return filas

def mostrar(filas, titulo):
    css = """<style>.biseccion{border-collapse:collapse;font-family:Arial;font-size:12px;width:100%;}
    .biseccion th{background:#183153;color:white;padding:8px;border:1px solid #cbd5e1;}
    .biseccion td{padding:6px;border:1px solid #cbd5e1;text-align:right;}
    .biseccion tr:nth-child(even){background:#f1f5f9;}
    .biseccion td.cambio{background:#fee2e2!important;color:#b91c1c;font-weight:bold;text-decoration:underline;}
    .biseccion td.ok{background:#dcfce7;color:#166534;font-weight:bold;}</style>"""
    cab = ['i','XL','XU','XR','f(XL)','f(XR)','f(XL) × f(XR)','EA (%)','Criterio de paro']
    partes = [css, '<h3>'+titulo+'</h3>', '<div style="overflow-x:auto"><table class="biseccion"><thead><tr>']
    partes += ['<th>'+h+'</th>' for h in cab]
    partes.append('</tr></thead><tbody>')
    for i,xl,xu,xr,fxl,fxr,p,ea,paro,cambio in filas:
        vals=[i,xl,xu,xr,fxl,fxr,p,ea,paro]
        partes.append('<tr>')
        for j,v in enumerate(vals):
            clase = 'cambio' if (j==1 and cambio=='XL') or (j==2 and cambio=='XU') else ('ok' if j==8 and paro else '')
            texto = '—' if v is None else (f'{v:.10f}' if isinstance(v,float) else str(v))
            partes.append(f'<td class="{clase}">{texto}</td>')
        partes.append('</tr>')
    partes.append('</tbody></table></div>')
    display(HTML(''.join(partes)))
    ultima = filas[-1]
    print(f'Raíz aproximada = {ultima[3]:.10f} | Iteraciones = {len(filas)} | Error = {ultima[7] if ultima[7] is not None else 0:.7f}%')
```

## 2. Ejercicio 1 — Función exponencial: f(x) = exp(-x) - x
**Intervalo inicial:** XL = -0.5, XU = 2 · **Error permitido:** 0,005 %.

```python
f1 = lambda x: math.exp(-x)-x
filas1 = biseccion(f1, -0.5, 2, tolerancia=0.005)
mostrar(filas1, "Ejercicio 1 — Función exponencial: f(x) = exp(-x) - x")
```

## 3. Ejercicio 2 — Primera raíz: f(x) = 4x⁴ - 5x² + 10x - 12
**Intervalo inicial:** XL = -2.5, XU = -1 · **Error permitido:** 0,005 %.

```python
f2 = lambda x: 4*x**4-5*x**2+10*x-12
filas2 = biseccion(f2, -2.5, -1, tolerancia=0.005)
mostrar(filas2, "Ejercicio 2 — Primera raíz: f(x) = 4x⁴ - 5x² + 10x - 12")
```

## 4. Ejercicio 3 — Segunda raíz: f(x) = 4x⁴ - 5x² + 10x - 12
**Intervalo inicial:** XL = 0.5, XU = 2 · **Error permitido:** 0,005 %.

```python
f3 = lambda x: 4*x**4-5*x**2+10*x-12
filas3 = biseccion(f3, 0.5, 2, tolerancia=0.005)
mostrar(filas3, "Ejercicio 3 — Segunda raíz: f(x) = 4x⁴ - 5x² + 10x - 12")
```

## ¿Cómo explicar el programa?

En cada iteración se calcula `XR = (XL + XU)/2`. Después se evalúa `f(XL)*f(XR)`:

- **Negativo:** se reemplaza XU por XR.
- **Positivo:** se reemplaza XL por XR.
- **Cero:** se encontró la raíz exacta.

Luego se calcula el **error aproximado porcentual** y se repite hasta que sea menor que **0,005 %**. Los valores que cambiaron están **en rojo y subrayados**. En Colab puedes modificar los intervalos iniciales y volver a ejecutar.
